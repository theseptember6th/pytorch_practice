## Topic 4: Creating Tensors
- Here are the most important methods to create tensors.

```python
import torch
# 1. torch.empty(): Allocates memory without initializing (values are whatever was in memory).
a = torch.empty(2, 3) # Creates a 2x3 tensor
print("torch.empty:\n", a)
# 2. torch.zeros(): Creates a tensor filled with zeros.
b = torch.zeros(2, 3)
print("\ntorch.zeros:\n", b)
# 3. torch.ones(): Creates a tensor filled with ones.
c = torch.ones(2, 3)
print("\ntorch.ones:\n", c)
# 4. torch.rand(): Creates a tensor with random values between 0 and 1.
d = torch.rand(2, 3)
print("\ntorch.rand:\n", d)
# 5. For reproducibility with random numbers, set a seed.
torch.manual_seed(100)
e = torch.rand(2, 3)
print("\ntorch.rand with seed 100:\n", e)
# 6. torch.tensor(): Creates a tensor from a Python list or NumPy array.
f = torch.tensor([[4, 5, 6], [7, 8, 9]])
print("\ntorch.tensor from list:\n", f)
# 7. Other useful methods
g = torch.arange(0, 10, 2) # torch.arange(start, end, step)
print("\ntorch.arange:\n", g)
h = torch.linspace(0, 10, 5) # torch.linspace(start, end, steps)
print("\ntorch.linspace:\n", h)
i = torch.eye(3) # Identity matrix
print("\ntorch.eye:\n", i)
j = torch.full((2, 3), 5) # Tensor of shape 2x3 filled with value 5
print("\ntorch.full:\n", j)
```

## Topic 5: Tensor Shape and Data Type

```python
x = torch.tensor([[4, 5, 6], [7, 8, 9]])
# 1. Getting shape
print("Shape of x:", x.shape) # or x.size()
print("Number of dimensions (rank):", x.dim())
print("Total number of elements:", x.numel())
# 2. Getting data type
print("Data type of x:", x.dtype)
# 3. Specifying data type during creation
int_tensor = torch.tensor([1, 2, 3], dtype=torch.int32)
float_tensor = torch.tensor([1, 2, 3], dtype=torch.float32)
print("\nInteger Tensor:", int_tensor)
print("Float Tensor:", float_tensor)
# 4. Changing data type of an existing tensor
x_float = x.to(torch.float32)
print("\nOriginal x dtype:", x.dtype)
print("New x_float dtype:", x_float.dtype)
# 5. Using the `like` functions
y = torch.rand_like(x_float) # Creates a random tensor of same shape as x_float
print("\nrand_like (float):\n", y)
```

## 6.1 Scalar Operations
- (Element wise operation with single tensor)
- Operations between a tensor and a single number.

```python
x = torch.tensor([1, 2, 3, 4], dtype=torch.float32)
print("Original x:", x)
print("x + 2:", x + 2)
print("x - 2:", x - 2)
print("x * 2:", x * 2)
print("x / 2:", x / 2)
print("x ** 2 (power):", x ** 2)
```

## 6.2 Element-wise Operations
- Operations between two tensors of the same shape.


```python
a = torch.tensor([1, 2, 3])
b = torch.tensor([4, 5, 6])
print("a + b:", a + b)
print("a * b:", a * b)
print("a / b:", a / b)
print("a //b :",a//b)
```

## Topic 7: In-Place Operations (The _ Suffix)
- In-place operations modify the original tensor directly, saving memory

```python
m = torch.tensor([1, 2, 3])
n = torch.tensor([4, 5, 6])
# Non-in-place: creates a new tensor
result = m + n# will return new tensor of same size,and this may be bad if we r having storage issue,and the tensors are very large
print("m after non-in-place addition:", m) # m is unchanged
# In-place: modifies m directly
m.add_(n) # Equivalent to m = m + n, but modifies m in-place # m is taking the sum value in its own position
print("m after in-place addition (add_):", m)
```

```python
# In-place ReLU activation
a = torch.tensor([-1, 2, -3, 4])
a.relu_() # Replaces negative values with 0
print("\na after in-place ReLU (relu_):", a)
```

## Topic 8: Copying Tensors (Clone vs. Assignment)
- Assignment (=): Creates a new reference to the same memory location. Changes to one
affect the other.
-  clone(): Creates a new, independent copy in a new memory location

```python
a = torch.tensor([1, 2, 3])
print("Original a:", a)
# Shallow copy using assignment
b = a
print("b = a, b:", b)
# Modify a
a[0] = 100
print("\na after change:", a)
print("b after a changed:", b) # b changes because it points to the same memory
# Deep copy using clone
c = a.clone()
a[1] = 200
print("\na after further change:", a)
print("c after a changed (clone):", c) # c remains unchanged
print("\nMemory IDs:")
print("ID of a:", id(a))
print("ID of b (assigned):", id(b)) # Same as a
print("ID of c (cloned):", id(c)) # Different from a
```

## Topic 9: Moving Tensors to GPU

## checking on windows 

```python
# 1. Check if GPU is available
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Using device:", device)
# 2. Create a tensor directly on the GPU(need to specify device)
gpu_tensor = torch.rand(2, 3, device=device)
print("\nTensor created on GPU:\n", gpu_tensor)
print("Device of gpu_tensor:", gpu_tensor.device)
# 3. Move an existing CPU tensor to GPU
cpu_tensor = torch.rand(2, 3)# this is made on cpu(by default if device not specified)
print("\nCPU tensor device:", cpu_tensor.device)
gpu_tensor_2 = cpu_tensor.to(device)
print("GPU tensor device after .to():", gpu_tensor_2.device)
# 4. Speed comparison (example code)
if torch.cuda.is_available():
 import time
 size = 10000
 cpu_a = torch.rand(size, size)
 cpu_b = torch.rand(size, size)
 # CPU Computation
 start = time.time()
 cpu_result = torch.matmul(cpu_a, cpu_b)
 cpu_time = time.time() - start
 # GPU Computation
 gpu_a = cpu_a.to(device)
 gpu_b = cpu_b.to(device)
 start = time.time()
 gpu_result = torch.matmul(gpu_a, gpu_b) # Wait for GPU to finish
 torch.cuda.synchronize() # Ensure all GPU operations are complete
 gpu_time = time.time() - start
 print(f"\nCPU time for {size}x{size} matmul: {cpu_time:.2f}seconds")
 print(f"GPU time for {size}x{size} matmul: {gpu_time:.2f} seconds")
 print(f"Speedup: {cpu_time/gpu_time:.2f}x")
```

### Speed Comparison betn CPU and GPU

```python
import torch
import time

# define the size of the matrices
size=10000 # large size for performance comparison


#create random matrices on CPU
matrix_cpu1=torch.randn(size=(size,size))
matrix_cpu2=torch.randn(size=(size,size))

#Measure time on CPU
start_time=time.time()
result_cpi=torch.matmul(matrix_cpu1,matrix_cpu2) # matrix multiplication on CPU
cpu_time=time.time()-start_time

print(f"Time on CPU: {cpu_time:.4f} seconds")

#Move matrices on GPU
matrix_gpu1=matrix_cpu1.to("cuda")
matrix_gpu2=matrix_cpu2.to("cuda")

#Measure time on GPU
start_time=time.time()
result_gpu=torch.matmul(matrix_gpu1,matrix_gpu2)
torch.cuda.synchronize() # Ensure all GPU operations are complete
gpu_time=time.time()-start_time

print(f"Time on GPU : {gpu_time:.4f} seconds")


#Compare results
print(f"\n Speedup (CPU time/GPU time) : {cpu_time/gpu_time}")
```

## Topic 10: Reshaping Tensors

```python
# Create a tensor of shape (4, 4)
r = torch.arange(16).reshape(4, 4)
print("Original tensor (4x4):\n", r)
# 1. reshape()
reshaped = r.reshape(2, 8)
print("\nReshape to (2, 8):\n", reshaped)
# 2. flatten()
flat = r.flatten()
print("\nFlattened (1D):", flat)
# 3. permute() - reorders dimensions
# Create a 3D tensor (2, 3, 4)
p = torch.arange(24).reshape(2, 3, 4)
print("\nOriginal 3D shape (2,3,4):", p.shape)
# Permute dimensions: move dimension 2 (index 2) to front, then 0,then 1
permuted = p.permute(2, 0, 1)
print("Permuted shape (4,2,3):", permuted.shape)
# 4. unsqueeze() - adds a dimension of size 1
img = torch.rand(226, 226, 3) # Example image shape
print("\nImage shape:", img.shape)
batch_img = img.unsqueeze(0) # Add batch dimension at position 0
print("After unsqueeze (batch added):", batch_img.shape) # (1, 226,226, 3)
# 5. squeeze() - removes dimensions of size 1
batch_removed = batch_img.squeeze(0)
print("After squeeze (batch removed):", batch_removed.shape)
```

## Topic 11: Interoperability with NumPy and torch

```python
import numpy as np
# 1. PyTorch Tensor -> NumPy Array
tensor = torch.tensor([1, 2, 3])
print("PyTorch Tensor:", tensor)
numpy_array = tensor.numpy()
print("Converted NumPy Array:", numpy_array)
print("Type:", type(numpy_array))


# 2. NumPy Array -> PyTorch Tensor
np_array = np.array([4, 5, 6])
print("\nNumPy Array:", np_array)
tensor_from_np = torch.from_numpy(np_array)
print("Converted PyTorch Tensor:", tensor_from_np)
print("Type:", type(tensor_from_np))
```
