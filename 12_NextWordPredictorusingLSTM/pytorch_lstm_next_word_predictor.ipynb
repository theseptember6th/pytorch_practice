```python
#Step 1: Tokenize the entire document using nltk's word_tokenize
tokens = word_tokenize(document.lower())
print(f"All tokens : {tokens[:20]}") # first 20 tokens ,there may contain repetition of same token
```

```python
# Step2 : Create vocabulary with unique tokens
counter=Counter(tokens) # counter creates dictionary of each word in that list  and  value as their total counts for each word
unique_tokens=list(counter.keys())
print(f"Number of unique tokens: {len(unique_tokens)}")
print(unique_tokens[:20])
```

```python
## Step 3: Build word to -index dictionary
vocab={"<UNK>":0} #unknown token for words not in vocabulary
for token in unique_tokens:
    vocab[token]=len(vocab) # trick to add keys and values in that dicitionary ,use ur brain


print(f"Vocabulary size: {len(vocab)}")
print(f"Some vocab entries : {list(vocab.items())[:10]}") # just temporarily converting dict to list and then checking first 10
```

```python
## Step 1: Extract sentences from document
input_sentences=document.split("\n") # assuming sentences separated by new line
print(f"Number of sentences: {len(input_sentences)}") # total sentences in that document
print(f"First Sentence: {input_sentences[0]}")
```

```python
## Step2 : Function to convert those above sentences text to indices/numerical using vocabulary we defined earlier. eg. one sentence [1,4,5,...]

def text_to_indices(sentence,vocab):
    """Convert a tokenized sentence to list of indices using vocabulary"""
    numerical_sentence=[]
    for token in sentence:
        if token in vocab:
            numerical_sentence.append(vocab[token])
        else:
            numerical_sentence.append(vocab["<UNK>"])
    
    return numerical_sentence


##Step 3:Now use that FUnction to  Convert all sentences  text to indices/nuemrical form 
input_numerical_sentences=[]
for sentence in input_sentences:
    if sentence.strip(): # skip empty sentences
        #tokenize the sentence
        tokenized_sentence=word_tokenize(sentence.lower())
        #convert to indices
        numerical_sentence=text_to_indices(tokenized_sentence,vocab)

        input_numerical_sentences.append(numerical_sentence)

print(f"First Sentence: {input_sentences[0]}")
print(f"First sentence as indices/numbers: {input_numerical_sentences[0]}")
print(f"length of first sentence: {len(input_numerical_sentences[0])}")
```

## Code for Creating Training Sequences

```python
# Creating training sequences from all sentences
training_sequences=[]

for sentence in input_numerical_sentences:
    for i in range(1,len(sentence)):
        #Input:From start upto i (not including i)
        input_seq=sentence[:i]
        #Output:ith word (index i)
        output_seq=sentence[i]

        training_sequences.append(input_seq)

print(f"Total training sequence created: {len(training_sequences)}")

#Display first 5 sentences
for i,seq in enumerate(training_sequences[:5]):
    print(f"Sequence {i+1}: {seq}")
```

## Code for padding

```python
## Step1 : Find the longest sequence
length_list=[len(seq) for seq in training_sequences]
max_len=max(length_list)
print(f"Longest sequence length : {max_len}")

## Step 2: Padd all sequences to max_len
padded_sequences=[]

for seq in training_sequences:
    #Calculate how many zeros needed
    padding_needed=max_len - len(seq)
    #Create padding list of zeros
    padding=[0]* padding_needed
    #combine padding + original sequence (pre-padding)
    padded_seq=padding+seq
    
    padded_sequences.append(padded_seq)

print(f"First sentence after padding: {padded_sequences[0]}")
print(f"Length after padding: {len(padded_sequences[0])}")


## Step 3: Convert that  to pytorch tensor
padded_sequences=torch.tensor(padded_sequences,dtype=torch.long)
print(f"Shape of padded sequences: {padded_sequences.shape}")
```

## Code for Dataset and DataLoader

```python
from torch.utils.data import Dataset,DataLoader

#custom Dataset class

class CustomDataset(Dataset):
    def __init__(self,X,y):
        self.X=X
        self.y=y
    
    def __len__(self):
        return len(self.X)

    def __getitem__(self,idx):
        return self.X[idx],self.y[idx]

    
#Create dataset object
dataset=CustomDataset(X,y)
print(f"Dataset length: {len(dataset)}")

#Create dataloader
batch_size=32
dataloader=DataLoader(dataset,batch_size=batch_size,shuffle=True)

print(f"Numer of batches: {len(dataloader)}")

#Test the dataloader

for batch_X,batch_y in dataloader:
    print(f"Batch X shape : {batch_X.shape}")
    print(f"Batch y shape : {batch_y.shape}")
    break
```

## Code for LSTM Model

```python
import torch.nn as nn

class LSTMModel(nn.Module):

    def __init__(self,vocab_size,embedding_dim=100,hidden_dim=150):
        super().__init__()
    
        #Embedding layer : converts word indices to dense vectors
        self.embedding=nn.Embedding(vocab_size,embedding_dim)

        #LSTM layer
        self.lstm=nn.LSTM(
            input_size=embedding_dim, #size of each input vector
            hidden_size=hidden_dim, #Size of hidden state
            batch_first=True  #Input shape: (batch,seq,feature)
        )

        #FUlly connected layer for output
        self.fc=nn.Linear(hidden_dim,vocab_size)



    def forward(self,x):
        # x shape : (batch_size,seq_len)

        #step1 : Embedding
        embedded=self.embedding(x)
        #embedded shape: (batch_size,seq_len,embedding_dim)

        #step 2: Pass through LSTM
        #lstm_out: all hidden states for each timestep
        #(hidden,cell):final hidden and cell states

        lstm_out,(hidden,cell)=self.lstm(embedded)
        #lstm_out shape:(batch_size,seq_len,hidden_dim)
        #hidden shape: (1,batch_size,hidden_dim)

        #step 3:Take final hidden state(last time step)
        final_hidden=hidden[-1] #Remove the 1 from dimension
        #final_hidden shape: (batch_size,hidden_dim)


        #Step 4:pass through linear layer
        output=self.fc(final_hidden)
        #output shape: (batch_size,vocab_size)

        return output



```

```python
## Create model instance
vocab_size=len(vocab)
model=LSTMModel(vocab_size=vocab_size,embedding_dim=100,hidden_dim=150)
model.to(device)
print(model)

#Count parameters
total_params=sum(p.numel() for p in model.parameters())
print(f"Total parameters: {total_params}")
```

## Code for training

```python
import torch.optim as optim

#Hyperparameters
epochs=50
learning_rate=0.001

#loss function and optimizer
criterion=nn.CrossEntropyLoss()
optimizer=optim.Adam(model.parameters(),lr=learning_rate) #add wegiht_decay for l2 regularization


#training loop
print("Start Training ...")
for epoch in range(epochs):
    # set model to train mode
    model.train()
    total_loss=0

    for batch_X,batch_y in dataloader:
        #Move data to device
        batch_X=batch_X.to(device)
        batch_y=batch_y.to(device)

        #Reset gradients
        optimizer.zero_grad()

        #Forward pass
        outputs=model(batch_X)

        #calculate loss
        loss=criterion(outputs,batch_y) #(predicted,true)

        #backward pass
        loss.backward()

        #Update weights
        optimizer.step()

        #Accumulate loss
        total_loss+=loss.item()

    
    #Print progess every epoch
    avg_loss=total_loss/len(dataloader)
    print(f"Epoch{epoch+1}/{epochs},loss:{avg_loss:.4f}")

print("Training Complete")
```

## Code for Prediction

```python
def predict_next_word(model,vocab,text,max_len=10):
    """
    Predict the next word given a text input
    """

    model.eval() #set model to evaluation mode

    #Step1: tokenize the input text
    tokenized_text=word_tokenize(text.lower())
    print(f"Tokenized: {tokenized_text}")

    #Step2:Convert to indices
    numerical_text=[]
    for token in tokenized_text:
        if token in vocab:
            numerical_text.append(vocab[token])
        else:
            numerical_text.append(vocab["<UNK"])
    print(f"indices : {numerical_text}")

    #Step3: pad the sequence
    #Note:we neeed max_len -1 becase we will add the prediction
    padding_needed=(max_len-1) - len(numerical_text)

    if padding_needed > 0:
        padding=[0]* padding_needed
        padded_text=padding + numerical_text
    else:
        padded_text=numerical_text[:max_len-1] #Truncate if too long


    print(f"Padded:{padded_text}")


    #Step 4: convert to tensor
    input_tensor=torch.tensor(padded_text,dtype=torch.long).unsqueeze(0) #add batch dimension
    input_tensor=input_tensor.to(device)

    #Step 5: Get prediction by sending it to model
    with torch.no_grad():
        output=model(input_tensor)
    
    #Step 6:get index with highest probability
    
    predicted_idx=torch.argmax(output,dim=1).item()


    # Step 7: convert the index back to word
    # reverse mapping : index -> word
    idx_to_word={idx:word for word,idx in vocab.items()}
    predicted_word=idx_to_word.get(predicted_idx,"<UNK>")
    
    return predicted_word
```

```python
import time
def generate_text(model, vocab, seed_text, num_words=10, max_len=10):
    """ Generate multiple words sequentially """
    model.eval()
    generated_text=seed_text
    current_text=seed_text

    print(f"Genrating {num_words} words from: '{seed_text}'")
    print("-"*50)

    for i in range(num_words):
        #Predict next word
        next_word=predict_next_word(model,vocab,current_text,max_len)

        #Append to generated text
        generated_text+=" "+ next_word
        current_text=current_text+" "+next_word

        #Print word by word (with delay for effect)
        print(f"Word {i+1}: '{next_word}'")
        time.sleep(0.5) #0.5 second dealay

    
    print("-"*50)
    print(f"Completed generated text:\n {generated_text}")
    return generated_text
```

```python
def predict_top_k(model, vocab, text, k=3, max_len=10):
    """
    Predict top k most likely next words
    """
    model.eval()

    # Same preprocessing as before
    tokenized_text = word_tokenize(text.lower())
    numerical_text = []
    for token in tokenized_text:
        if token in vocab:
            numerical_text.append(vocab[token])
        else:
            numerical_text.append(vocab['<UNK>'])

    # Padding
    padding_needed = (max_len - 1) - len(numerical_text)
    if padding_needed > 0:
        padding = [0] * padding_needed
        padded_text = padding + numerical_text
    else:
        padded_text = numerical_text[:max_len-1]

    # Convert to tensor and move to device
    input_tensor = torch.tensor(padded_text, dtype=torch.long).unsqueeze(0)
    input_tensor = input_tensor.to(device)

    # Get prediction
    with torch.no_grad():
        output = model(input_tensor)

    # Apply softmax to get probabilities
    probabilities = torch.softmax(output, dim=1)

    # Get top k indices and values
    top_k_probs, top_k_indices = torch.topk(probabilities, k, dim=1)

    # Convert to words
    idx_to_word = {idx: word for word, idx in vocab.items()}
    suggestions = []
    for i in range(k):
        word = idx_to_word.get(top_k_indices[0][i].item(), '<UNK>')
        prob = top_k_probs[0][i].item()
        suggestions.append((word, prob))

    return suggestions

# Test top-k predictions
test_text = "you will have to fill "
suggestions = predict_top_k(model, vocab, test_text, k=3)
print(f"Input: '{test_text}'")
print("Suggestions:")
for word, prob in suggestions:
    print(f" - {word} ({prob:.2%})")
```
