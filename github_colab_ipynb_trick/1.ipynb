The absolute fastest way to open a GitHub `.ipynb` file in Google Colab—completely temporarily and without saving it to your Google Drive—is by using a quick URL tweak.

Here is the shortcut and how it works.

### The Quick URL Swap

You just need to replace `github.com` in your browser's address bar with `colab.research.google.com/github`.

**The Formula:**

* **Original GitHub URL:** `https://github.com/user/repository/blob/main/notebook.ipynb`
* **Modified Colab URL:** `https://colab.research.google.com/github/user/repository/blob/main/notebook.ipynb`

---

### Why this keeps your Drive clean

When you open a notebook directly from GitHub using this method:

* **It runs in a sandbox:** Colab loads the file into a temporary session.
* **No auto-save:** Any changes you make or cells you run will **not** be saved to your Google Drive automatically.
* **How to discard:** When you are done, simply close the browser tab. The notebook disappears from your Colab history, leaving your Google Drive completely untouched.



## Best solution

That makes complete sense—`Cmd + D` (or `Ctrl + D`) is deeply hardwired into browsers for bookmarking, so we don't want to mess with that.

Instead of an actual keyboard shortcut, the most efficient, native way to do this in **one click** without installing heavy extensions is using a **Bookmarklet**. It sits right in your bookmarks bar, and clicking it instantly transforms the URL.

Here is exactly how to set it up in less than a minute.



### Create the "1-Click" Button

1. Right-click any empty space on your browser's bookmarks bar and select **Add Page...** (or **Add Bookmark...**).
2. Set the **Name** to something short like `🚀 Open in Colab`.
3. In the **URL** (or **Location**) field, delete whatever text is there and paste this exact block of code:

```javascript
javascript:(function(){var url=window.location.href;if(url.includes('github.com')){window.open(url.replace('github.com','colab.research.google.com/github'),'_blank');}else{alert('You are not on a GitHub page!');}})();

```

4. Click **Save**.

### How to use it

Whenever you are browsing a public repository on GitHub and looking at a `.ipynb` file:

1. Simply click the `🚀 Open in Colab` button on your bookmarks bar.
2. It will instantly open that exact notebook in Google Colab in a **new tab**, leaving your current GitHub tab exactly where it was.

Since it uses standard browser behavior, it bypasses the need for keyboard shortcuts entirely, doesn't background-run any extension scripts, and keeps your temporary workspace completely isolated from your Google Drive.
